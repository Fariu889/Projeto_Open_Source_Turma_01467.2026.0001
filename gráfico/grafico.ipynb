```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("tips")
df.head()
```

```python
tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "day"
)
plt.show()
```

```python
fmri = sns.load_dataset("fmri")

sns.lineplot(
    data = fmri,
    x = "timepoint",
    y = "signal",
    hue = "event"
)
plt.show()
```

```python
titanic = sns.load_dataset("titanic")

sns.barplot(
    data = titanic,
    x = "class",
    y = "fare",
    palette = "viridis"
)

plt.show()
```

```python
iris  = sns.load_dataset("iris")

sns.histplot(
    data = iris,
    x = "sepal_length",
    kde= True,
    bins = 15
)
plt.show()
```

```python
corr = iris.corr(numeric_only= True)

sns.heatmap(
    corr,
    annot = True,
    cmap= "coolwarm"
)
plt.show()
```
