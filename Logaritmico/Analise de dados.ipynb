```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

tabela = pd.read_excel("atividade_transformacao_logaritmica.xlsx")
tabela
```

```python
visitas_regiao = tabela.groupby("Regiao")["Visitas_Mensais"].sum().reset_index(name= "Visitas_Mensais")
visitas_regiao
```

```python
sns.lineplot(
    data = visitas_regiao,
    x = "Regiao",
    y = "Visitas_Mensais"
)

plt.show()
```

```python
visitas_regiao["Visitas_Mensais"] = np.log1p(visitas_regiao["Visitas_Mensais"])

visitas_regiao
```

```python
sns.lineplot(
    data = visitas_regiao,
    x = "Regiao",
    y = "Visitas_Mensais"
)

plt.show()
```

```python
nordeste = tabela[tabela["Regiao"] == "Nordeste"]

nordeste
```

```python
nordeste = nordeste.sort_values("Visitas_Mensais")
nordeste
```

```python
sns.boxplot(
    data = nordeste,
    y = "Visitas_Mensais",
    x = "Canal_Principal",
    hue = "Canal_Principal"
)
plt.show()
```

```python
nordeste["Visitas_Mensais"] = np.log1p(nordeste["Visitas_Mensais"])

nordeste
```

```python
plt.title("Visitas mensais do Nordeste")
sns.boxplot(
    data = nordeste,
    y = "Visitas_Mensais",
    x = "Canal_Principal",
    hue = "Canal_Principal"
)
plt.ylabel("Visitas")
plt.xlabel("Canais")
plt.show()
```
