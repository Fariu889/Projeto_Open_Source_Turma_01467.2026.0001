```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day"
)

plt.show()
```

```python
fmri = sns.load_dataset("fmri")
sns.lineplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event"
)

plt.show()
```

```python
titanic = sns.load_dataset("titanic")
sns.barplot(
    data=titanic,
    x="class",
    y="fare",
    palette="viridis"
)

plt.show()
```

```python
iris = sns.load_dataset("iris")

sns.histplot(
    data=iris,
    x="sepal_length",
    kde=True,
    bins=15
)
plt.show()
```

```python
iris = sns.load_dataset("iris")
corr = iris.corr(numeric_only=True)
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
    
)
plt.show()
```

```python

sns.set_theme (style="darkgrid")

plt.figure(figsize=(8, 4))

ax = sns.barplot (x="day", y="total_bill", data=tips)

ax.set_title("Faturamento Médio por Dia",fontsize=14)

ax.set_xlabel("Dia da Semana")

ax.set_ylabel("Total da Conta ($)")

plt.show()
```
