```python
import pandas as pd
import glob

emprestimos = glob.glob('Datasets/dados_emprestimos/*.csv')
```

```python
lista_emprestimos = []

for arquivo in emprestimos:
    df_emprestimo = pd.read_csv(arquivo)
    lista_emprestimos.append(df_emprestimo)

df_emprestimo = pd.concat(lista_emprestimos, ignore_index = True)
```

```python
df_emprestimo['data_emprestimo'] = pd.to_datetime(df_emprestimo['data_emprestimo']).dt.strftime('%d/%m/%Y')
df_emprestimo['data_renovacao'] = pd.to_datetime(df_emprestimo['data_renovacao']).dt.strftime('%d/%m/%Y')
df_emprestimo['data_devolucao'] = pd.to_datetime(df_emprestimo['data_devolucao']).dt.strftime('%d/%m/%Y')


```

```python
df_emprestimo['matricula_ou_siape'] = df_emprestimo['matricula_ou_siape'].astype(str)

df_emprestimo
```
